```python
import random
import matplotlib.pyplot as plt
from collections import Counter
```

```python
def simulate_esp_test(num_trials=20, num_symbols=5):
    """
    Simulate one ESP test where someone guesses symbols.
    Returns the number of correct guesses out of num_trials.
    """
    correct_guesses = 0
    for _ in range(num_trials):
        # True symbol (what the "sender" is thinking)
        true_symbol = random.randint(1, num_symbols)
        # Guess (random, since no actual ESP)
        guess = random.randint(1, num_symbols)
        
        if true_symbol == guess:
            correct_guesses += 1
    
    return correct_guesses
```

```python
def run_simulation(num_experiments=300000000, trials_per_experiment=20, num_symbols=5):
    results = []
    for _ in range(num_experiments):
        correct = simulate_esp_test(trials_per_experiment, num_symbols)
        results.append(correct)
    return results
```

```python
# Run the simulation
print("Simulating ESP tests...")
print("Each test: 20 guesses with 5 possible symbols (20% chance by luck)")
print("Running 300000000 simulated experiments...\n")

results = run_simulation(300000000, 20, 5)

# Calculate statistics
total_experiments = len(results)

# Show the most extreme results
max_correct = max(results)
min_correct = min(results)
print(f"Most extreme results:")
print(f"Highest score: {max_correct}/20")
print(f"Lowest score: {min_correct}/20")

# Count frequency of each score
score_counts = Counter(results)
print("\nDistribution of scores:")
for score in sorted(score_counts.keys()):
    count = score_counts[score]
    percentage = count / total_experiments * 100
    bar = "█" * int(percentage / 2)  # Visual bar
    print(f"{score:2d} correct: {count:4d} times ({percentage:4.1f}%) {bar}")

# Create histogram
plt.figure(figsize=(12, 6))
plt.hist(results, bins=range(0, max(results)+2), alpha=0.7, color='skyblue', edgecolor='black')
plt.axvline(x=4, color='red', linestyle='--', label='Expected (4 correct)')
plt.axvline(x=8, color='orange', linestyle='--', label='8 correct')
plt.xlabel('Number of Correct Guesses (out of 20)')
plt.ylabel('Frequency')
plt.title('Distribution of ESP Test Results (10,000 Simulations)\nEach test: 20 guesses, 5 symbols')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
