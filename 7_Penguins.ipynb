## The K-Means Algorithm: How Does It Work?

We can spot possible clusters with our eyes — but how does a computer find them on its own? K-Means uses a simple, repeating process:

<img src = 'https://github.com/jdomyancich/big-data-camp/blob/main/imgs/kmeans.gif?raw=true'>

1.  **Choose 'k'** — how many clusters to look for. We know there are 3 species, so **k=3**.
2.  **Place centroids** — drop k center-points (the X's above) randomly onto the plot.
3.  **Assign** — attach every penguin to its nearest centroid, forming clusters.
4.  **Update** — move each centroid to the average position of its cluster's points.
5.  **Repeat** steps 3–4 until the centroids stop moving.

The algorithm finds the groups without ever being told the actual species!
