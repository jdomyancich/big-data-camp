## Grouping with `groupby()`

`groupby()` lets you split your data into groups, apply a function to each group, and combine the results — the "split-apply-combine" strategy, one of the most-used tools in Pandas.

### How it Works (The "Split-Apply-Combine" Strategy):

1. **Split** the data into groups that share the same value for the grouping key.
2. **Apply** an aggregation function to each group — `mean()`, `sum()`, `count()`, `max()`, `min()`, etc.
3. **Combine** the results back into one series or dataframe, with the grouping keys as the index.

All in a single line:

`df.groupby('split_group')[data_to_be_applied].function_applied_to_data`
