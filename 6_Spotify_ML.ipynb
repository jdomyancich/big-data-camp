### **1. Understanding the Data**

Spotify rates each song on a set of numeric **features**. A few we'll use:

* **`danceability`** — how suitable the track is for dancing (0–1).
* **`energy`** — intensity and activity (0–1).
* **`acousticness`** — how acoustic the track is (0–1).
* **`valence`** — musical positivity, sad → happy (0–1).
* **`liveness`** — likelihood it was performed live.
* **`instrumentalness`** — likelihood it has no vocals.
* **`speechiness`** — how much spoken word it contains.

Our goal: use these features to predict whether a song is **country** or **rap**.

`max_depth`: The maximum number of "if-then" questions the tree can ask in a row.
